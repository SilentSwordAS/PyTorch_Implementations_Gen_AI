```python
# Importing Necessary Libraries
import torch
import numpy as np
```

```python
# Converting Lists to Tensors
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
```

```python
# Converting Numpy Arrays to Tensors
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

```python
# Using factory methods to create tensors
x_ones = torch.ones_like(x_data)
print(x_ones)

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(x_rand)
```

```python
# Specifying shape for factory methods

shape = (2,3)

x_ones = torch.ones(shape)
print(x_ones)

x_rand = torch.rand(shape)
print(x_rand)

x_zeros = torch.zeros(shape)
print(x_zeros)
```

```python
# Attributes of a Tensor

print(x_data.shape)
print(x_data.dtype)
print(x_data.device)
```

```python
# Cat operation for joining Tensors
tensor = torch.ones(4,4)

# Column Wise Concatenation
print(torch.cat([tensor,tensor,tensor],dim=1))

# Row Wise Concatenation
print(torch.cat([tensor,tensor,tensor],dim=0))
```

```python
# Matrix Multiplication

y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

m3 = torch.rand_like(y1)
y3 = torch.matmul(tensor, tensor.T, out=m3)

print(y1)
print(y2)
print(y3)
```
